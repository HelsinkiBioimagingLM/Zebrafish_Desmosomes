```python
from cellpose import models, io, utils
from bioio import BioImage
import napari
```

```python
test_img = r"E:\KristinG\20260625_zebrafish\fin\DKO_ctrl_1_skin_1_ventral_fin_Out.czi"
img = BioImage(test_img)
img_rd = img.get_image_data("ZYX", C=1)
```

```python
anisotropy = img.physical_pixel_sizes[0]/img.physical_pixel_sizes[1]
```

```python
cp3 = io.logger_setup()
model = models.Cellpose(gpu=True, model_type='cyto3')
channels = [[0, 0]]  
masks, flows, styles, diams = model.eval(img_rd, diameter=200, flow_threshold=0.4, channels=channels, do_3D=True, z_axis=0, normalize=True, anisotropy=anisotropy)
```

```python
v = napari.Viewer()
```

```python
v.add_image(img_rd, name='image', colormap='gray',scale=(img.physical_pixel_sizes[0], img.physical_pixel_sizes[1], img.physical_pixel_sizes[2]))
v.add_labels(masks, name='masks norm anisotropy', opacity=0.5,scale=(img.physical_pixel_sizes[0], img.physical_pixel_sizes[1], img.physical_pixel_sizes[2]))
```
